```python
marks = []
n = int(input("how many marks?"))

for i in range(n):
    value = int(input("enter mark:"))
    marks.append(value)

if len(marks) > 0:
    average = sum(marks)/ len(marks)
    print("average:", average)
    print("highest:", max(marks))
    print("lowest:", min(marks))
else:
    print("no marks entered")
```

write a function that takes a list of numbers and returns how many are above 50

```python
def count_above_50(numbers):
    count = 0
    for num in numbers:
        if num > 50:
            count +=1
    return count
numbers = [20,60,45,80,30,70]
print(count_above_50(numbers))
```
